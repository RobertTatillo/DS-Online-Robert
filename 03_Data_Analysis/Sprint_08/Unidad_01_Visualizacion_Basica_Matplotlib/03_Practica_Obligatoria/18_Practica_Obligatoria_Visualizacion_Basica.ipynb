### Ejercicio 1

```python
df_cities = pd.read_csv("./data/california_cities.csv")
display(df_cities.head())
display(df_cities.describe())

# Quito las filas con nulos
df_cities = df_cities.dropna()
print("Filas tras quitar nulos:", df_cities.shape[0])
```

#### #1.2  

Recrea la siguiente figura, para ello construye en apartados separados cada gráfico y luego haz un apartado en el que se junten los cuatro.  

Las figuras son:
1. Las ciudades más pobladas, con los numeros de las poblaciones (tendrás que usar texto y recordar que para centrarlos se necesita poner el argumento `ha` a "center" o investigar como funciona bar_label). Los colores de las barras deben tener un 50% de transparencia y ser: azul, gris, rojo, aceituna y marrón.  

2. Histograma del área total ocupada por las ciudades. (50 bins, fuente = "Arial", tamaño = 14, peso = "Bold", color azul acero o "steelblue")
3. El scatter de área de agua de un ciudad con su poblacion. Transparencia al 50% y color verde.
4. El boxplot de la altura de las ciudades.

```python
# GRAFICO 1: las 5 ciudades mas pobladas
top5 = df_cities.sort_values("population_total", ascending=False).head(5)
colores = ["blue", "gray", "red", "olive", "brown"]

plt.figure(figsize=(9, 5))
barras = plt.bar(top5["city"], top5["population_total"], color=colores, alpha=0.5)
plt.bar_label(barras, fmt="%d")   # pone el numero encima de cada barra
plt.title("Ciudades mas pobladas")
plt.ylabel("Poblacion")
plt.xticks(rotation=45)
plt.show()
```

```python
# GRAFICO 2: histograma del area total
plt.figure(figsize=(9, 5))
plt.hist(df_cities["area_total_km2"], bins=50, color="steelblue")
plt.title("Area total ocupada por las ciudades", fontname="Arial", fontsize=14, fontweight="bold")
plt.xlabel("Area total (km2)")
plt.ylabel("Frecuencia")
plt.show()
```

```python
# GRAFICO 3: scatter area de agua vs poblacion
plt.figure(figsize=(9, 5))
plt.scatter(df_cities["area_water_km2"], df_cities["population_total"], color="green", alpha=0.5)
plt.title("Area de agua vs Poblacion")
plt.xlabel("Area de agua (km2)")
plt.ylabel("Poblacion")
plt.show()
```

```python
# GRAFICO 4: boxplot de la altura
plt.figure(figsize=(6, 5))
plt.boxplot(df_cities["elevation_m"])
plt.title("Altura (elevacion) de las ciudades")
plt.ylabel("Elevacion (m)")
plt.show()
```

```python
# LOS CUATRO JUNTOS en una sola figura 2x2
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

# 1. Barras
barras = axes[0, 0].bar(top5["city"], top5["population_total"], color=colores, alpha=0.5)
axes[0, 0].bar_label(barras, fmt="%d")
axes[0, 0].set_title("Ciudades mas pobladas")
axes[0, 0].tick_params(axis="x", rotation=45)

# 2. Histograma
axes[0, 1].hist(df_cities["area_total_km2"], bins=50, color="steelblue")
axes[0, 1].set_title("Area total", fontname="Arial", fontsize=14, fontweight="bold")

# 3. Scatter
axes[1, 0].scatter(df_cities["area_water_km2"], df_cities["population_total"], color="green", alpha=0.5)
axes[1, 0].set_title("Area de agua vs Poblacion")

# 4. Boxplot
axes[1, 1].boxplot(df_cities["elevation_m"])
axes[1, 1].set_title("Altura de las ciudades")

plt.tight_layout()
plt.show()
```
